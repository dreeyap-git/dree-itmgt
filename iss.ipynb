```python
import requests as rq
import matplotlib.pyplot as mp
import time
from datetime import datetime

url = 'https://api.wheretheiss.at/v1/satellites/25544'

timestamp_list = []
latitude_list = []
longitude_list = []

print("Tracking the ISS...")
for i in range(100):
    fetch = rq.get(url)
    loc = fetch.json()
    
    timestamp = datetime.utcfromtimestamp(loc['timestamp']).strftime('%Y-%m-%d %H:%M:%S')
    latitude = loc['latitude']
    longitude = loc['longitude']

    timestamp_list.append(timestamp)
    latitude_list.append(latitude)
    longitude_list.append(longitude)

    print(f"{i+1}/100 | timestamp: {timestamp} | lat: {latitude}, long: {longitude}")
    time.sleep(10)

mp.figure(figsize=(10, 4))
mp.plot(longitude_list, latitude_list, marker='o')
mp.title("Track the ISS")
mp.xlabel("Longitude")
mp.ylabel("Latitude")
mp.grid(True)
mp.show()
```

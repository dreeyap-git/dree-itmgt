```python
#Find a way to save the text stored at https://www.gutenberg.org/cache/epub/10/pg10.txtLinks to an external site. into a Python variable called `pg10`.

import urllib.request

url = 'https://www.gutenberg.org/cache/epub/10/pg10.txt'

response = urllib.request.urlopen(url)
pg10 = response.read().decode('utf-8')
```

```python
#How many lines does the string have?

lines = len(pg10.splitlines())
print("No. of Lines:", lines)
```

```python
#How many words does the string have? (where a word is any set of directly-adjacent alphabet characters)

word_count = 0
in_word = False

for char in pg10:
    if char.isalpha():
        if not in_word:
            word_count += 1
            in_word = True
    else:
        in_word = False

print("Word Count:", word_count)
```

```python
#How many times does the word "apostle", ignoring capitalization, appear in the string?

pg10 = pg10.lower()

find = "apostle"
find_count = 0
current_word = ""

for char in pg10:
    if char.isalpha():
        current_word += char
    else:
        if current_word == find:
            find_count += 1
        current_word = ""
        
if current_word == find:
    find_count += 1

print("No. of times 'apostle' appears:", find_count)
```
